```python
import unittest
import sys


class Monster:
    """A cute monster"""

    def __init__(self, name, size, basic_type='fire'):
        """Creates a new monster"""
        self.name = name
        self.size = size
        self.basic_type = basic_type.lower()


    def fight(self, monster2):
        """Two monsters fight each other. Returns the name of the monster winning
        or 'draw' otherwise.

        Monster battles have very specific rules:
        - Water wins over Fire
        - Electrical wins over Water
        - Plant wins over Electrical
        - Fire wins over Plant

        if both monsters fighting have the same type, bigger monster wins!
        if both monsters fighting have the same type and size, it's a draw

        Any other combination, it's a draw
        """
        # Resposta
        # WATER
        if self.basic_type=='water':

            if monster2.basic_type=='fire':
                winner = self.name


            elif monster2.basic_type=='electrical':
                winner = monster2.name


            elif monster2.basic_type=='water':

                if self.size != monster2.size:
                    if self.size > monster2.size:
                        winner = self.name
                    else:
                        winner = monster2.name

                else:
                  winner = 'draw'
              
            else: 
              winner = 'draw'

        # ELECTRICAL
        elif  self.basic_type=='electrical':
            # Win
            if monster2.basic_type=='water':
                winner = self.name

            # Lose
            elif monster2.basic_type=='plant':
                winner = monster2.name

            # Same type
            elif monster2.basic_type=='electrical':

                if self.size != monster2.size:
                    if self.size > monster2.size:
                        winner = self.name
                    else:
                        winner = monster2.name

                else:
                  winner = 'draw'
            # Other combination  
            else: 
              winner = 'draw'

        # FIRE
        elif  self.basic_type=='fire':
            # Win
            if monster2.basic_type=='plant':
                winner = self.name

            # Lose
            elif monster2.basic_type=='water':
                winner = monster2.name

            # Same type
            elif monster2.basic_type=='fire':

                if self.size != monster2.size:
                    if self.size > monster2.size:
                        winner = self.name
                    else:
                        winner = monster2.name

                else:
                  winner = 'draw'
            # Other combination  
            else: 
              winner = 'draw'                   

        # PLANT
        elif  self.basic_type=='plant':
            # Win
            if monster2.basic_type=='electrical':
                winner = self.name

            # Lose
            elif monster2.basic_type=='fire':
                winner = monster2.name

            # Same type
            elif monster2.basic_type=='plant':

                if self.size != monster2.size:
                    if self.size > monster2.size:
                        winner = self.name
                    else:
                        winner = monster2.name

                else:
                  winner = 'draw'
            # Other combination  
            else: 
              winner = 'draw'

        return winner
```

def determine_winner(monster1, monster2):
    type_advantages = {
        'water': {'fire': monster1.name, 'electrical': monster2.name, 'plant': 'draw', 'water': 'size'},
        'electrical': {'water': monster1.name, 'plant': monster2.name, 'fire': 'draw', 'electrical': 'size'},
        'fire': {'plant': monster1.name, 'water': monster2.name, 'electrical': 'draw', 'fire': 'size'},
        'plant': {'electrical': monster1.name, 'fire': monster2.name, 'water': 'draw', 'plant': 'size'}
    }

    if monster1.basic_type not in type_advantages or monster2.basic_type not in type_advantages:
        return 'draw'

    result = type_advantages[monster1.basic_type].get(monster2.basic_type, 'draw')

    if result == 'size':
        if monster1.size > monster2.size:
            return monster1.name
        elif monster1.size < monster2.size:
            return monster2.name
        else:
            return 'draw'
    else:
        return result
