```python
import numpy as np

def llegir_rmsd(fitxer_rmsd):

    #RESPOSTA
    with open(fitxer_rmsd) as file:
        array_rmsd = np.loadtxt(file)

    return array_rmsd

    
llegir_rmsd('rmsd_uoc.dat')
```

```python
def rmsd_pas(fitxer_rmsd, pas_dinamica=0):

    #RESPOSTA
    rmsd_array = llegir_rmsd(fitxer_rmsd)
    pas_rmsd = rmsd_array[pas_dinamica,:]

    return pas_rmsd

rmsd_pas('rmsd_uoc.dat', 1)
```

```python
# Autor: Nicolas Rougier
# Font: http://www.labri.fr/perso/nrougier/teaching/numpy.100/
import numpy as np
SIZE = 20
STEPS =10

def iterate(Z):
    # Count neighbours
    N = (Z[0:-2,0:-2] + Z[0:-2,1:-1] + Z[0:-2,2:] +
         Z[1:-1,0:-2]                + Z[1:-1,2:] +
         Z[2:  ,0:-2] + Z[2:  ,1:-1] + Z[2:  ,2:])

    # Apply rules
    birth = (N==3) & (Z[1:-1,1:-1]==0)
    survive = ((N==2) | (N==3)) & (Z[1:-1,1:-1]==1)
    Z[...] = 0
    Z[1:-1,1:-1][birth | survive] = 1
    return Z

# Creem un tauler amb cèl·lules vives o mortes de forma aleatòria
Z = np.random.randint(0,2,(SIZE, SIZE))

# Simulem durant els passos indicats
for i in range(STEPS):
    Z = iterate(Z)

# Mostrem el tauler en el pas final
print(Z)
```
