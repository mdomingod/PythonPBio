```python
import numpy as np

# Coordenades àtom A
atom_A = [-0.325,0.945,0]
# Coordenades àtom B
atom_B = [0,0,0]
# Coordenades àtom C
atom_C = [1,0,0]

def angle_atoms(atom_A,atom_B,atom_C,graus=False):

  # RESPOSTA
  atoms_A_B_C = np.array([atom_A, atom_B, atom_C])
  x = []
  y = []
  z = []

  for i in range(0,3):
    x.append(atoms_A_B_C[i,0])
    y.append(atoms_A_B_C[i,1])
    z.append(atoms_A_B_C[i,2])
  
  # 1. Find the vectors AB and BC
  AB = (x[1]-x[0], y[1]-y[0], z[1]-z[0])
  BC = (x[2]-x[1], y[2]-y[1], z[2]-z[1])

  # 2. Dot product
  ABdotBC = np.dot(AB, BC)

  # 3. Magnitude
  AB_npMtude = np.linalg.norm(AB)
  BC_npMtude = np.linalg.norm(BC)

  
  # 4. Calculate the angle
  phi = np.arccos(ABdotBC/(AB_npMtude*BC_npMtude))

  if graus:
    # Angle in Degrees
    phi = np.degrees(phi)
  
  return phi

angle_atoms(atom_A, atom_B, atom_C, graus=True)

```
