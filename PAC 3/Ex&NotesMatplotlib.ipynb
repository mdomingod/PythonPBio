```python
# importem els mòduls necessaris
import matplotlib
import numpy as np
import matplotlib.pyplot as plt

def es_digit(data):
    # Funció que troba si un valor d'una string és un digit o no, i només retorna
    # els valors corresponents a lletres
   
    #RESPOSTA
  import math

  list_n = []
  only_string = ''

  for v in data:
    try:
      num = int(v)
      list_n.append(num)
            
    except ValueError as ve:
      #print("you've entered", v)
      only_string += v
  
  return only_string


def classifica_classes_CAZymes(data):
  # Funció en que creem un diccionari en que les keys son el tipus de classes
  # que poden arribar a existir pels CAZymes i els valors es correspon a la
  # quantitat existent d'aquestes classes.
  CAZymes = {}
  keys = []
  with open(data, "r") as infile:
    for line in infile:
      #resposta
      lis_lin = line.split()
      col1_lis_lin = lis_lin[0]
      keys.append(col1_lis_lin[0:2])
      CAZymes[col1_lis_lin[0:2]] = keys.count(col1_lis_lin[0:2])
  
  return CAZymes

      
classifica_classes_CAZymes('cazy_data_Aspergillus.txt')
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Define the range and step size
x_range = np.arange(-2 * np.pi / 2, 2 * np.pi / 2 + 0.1, 0.1)
y_range = np.arange(-2 * np.pi / 2, 2 * np.pi / 2 + 0.1, 0.1)

# Generate grid of x and y values
x, y = np.meshgrid(x_range, y_range)

# Compute the function values
z = 4 * np.cos(x) + 3 * np.sin(y)

# Create a 3D plot
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# Plot the surface
surf = ax.plot_surface(x, y, z, cmap='viridis')

# Set labels and title
ax.set_xlabel('X (radians)')
ax.set_ylabel('Y (radians)')
ax.set_zlabel('f(x,y)')
ax.set_title('3D Surface Plot of f(x,y) = 4*cos(x) + 3*sin(y)')

# Add a color bar which maps values to colors
fig.colorbar(surf)

# Show the plot
plt.show()
```

Representeu també, el r.m.s.f corresponent per a cada aminoàcid.      

El root mean square fluctuation (r.m.s.f.) és una mesura numèrica que mesura la flexibilitat individual de diferents partícules al llarg d'una trajectòria de dinàmica molecular. Aquesta mesura normalment ens serveix per a trobar quins aminoàcids són els que contribueixen més en el moviment de la proteïna.  

Les posicions del vector dades_dinamica són les següents:       

Número aminoàcid        
- Valor màxim distància entre substrat i aminoàcid (Å)  
- Valor mínim distància entre substrat i aminoàcid (Å)      
- Valor mitjà distància entre substrat i aminoàcid (Å)  
- Valor r.m.s.f. (Å)

```python
from matplotlib import figure
%matplotlib inline
import matplotlib
import numpy as np
import matplotlib.pyplot as plt

dades_dinamica = [["155", 2.394, 1.892, 2.152, 1.478],
                  ["156", 8.198, 4.859, 6.380, 1.389],
                  ["159", 2.327, 1.863, 2.065, 1.535],
                  ["191", 2.501, 1.958, 2.232, 0.815],
                  ["195", 2.358, 1.959, 2.159, 0.959],
                  ["394", 2.305, 1.923, 2.117, 0.768]]

# Resposta
x_numAminoAcid = []
y1_maxDist = []
y2_minDist = []
y3_mitjaDist = []
y4_rmsf = []

for i in range(0,6):
    x_numAminoAcid.append(dades_dinamica[i][0])
    y1_maxDist.append(dades_dinamica[i][1])
    y2_minDist.append(dades_dinamica[i][2])
    y3_mitjaDist.append(dades_dinamica[i][3])
    y4_rmsf.append(dades_dinamica[i][4])


fig, ax = plt.subplots(2, 1)
fig.set_figheight(15)
fig.set_figwidth(15)
ax[0].plot(x_numAminoAcid, y1_maxDist, color='red', label='Distància Màxima')
ax[0].plot(x_numAminoAcid, y2_minDist, color='blue', label='Distància Mínima')
ax[0].plot(x_numAminoAcid, y3_mitjaDist, color='orange', label='Distància Mitjana')

ax[0].set_xlabel('Número aminoàcid')
ax[0].set_ylabel('Distància (A)')

ax[0].set_title('Distàncies substrat - aminoàcid')

ax[0].legend()



ax[1].bar(x_numAminoAcid, y4_rmsf)


ax[1].set_xlabel('Número aminoàcid')
ax[1].set_ylabel('rmsf (A)')

ax[1].set_title('rmsf (A)')

plt.show()



```
